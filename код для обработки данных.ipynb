```python
import pandas as pd

df = pd.read_excel(r"C:\Users\Амина\Documents\data for project.xls")

# Показать первые 5 строк датасета
print(df.head(5))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Построение гистограммы для переменной wage
plt.figure(figsize=(8, 6))
plt.hist(df['wage'], bins=50)
plt.title('Распределение переменной wage')
plt.xlabel('Заработная плата')
plt.ylabel('Частота')
plt.show()
```

```python
# Построение гистограммы для переменной work_hours
plt.figure(figsize=(8, 6))
plt.hist(df['work_hours'], bins=40)
plt.title('Distribution of work_hours')
plt.xlabel('Work hours')
plt.ylabel('Frequency')
plt.show()
```

```python
# Построение гистограммы для переменной educ
plt.figure(figsize=(8, 6))
plt.hist(df['educ'], bins=40)
plt.title('Распределение переменной educ')
plt.xlabel('Образование')
plt.ylabel('Частота')
plt.show()
```

```python
# Построение гистограммы для переменной education_level
df["education_level"] = df["education_level"].round(2)
plt.figure(figsize=(8, 6))
df['education_level'].value_counts().plot(kind='bar')
plt.title('Распределение переменной education_level')
plt.xlabel('Уровень образования')
plt.ylabel('Частота')
plt.show()
```

```python
# Построение гистограммы для переменной age
plt.figure(figsize=(8, 6))
plt.hist(df['age'], bins=40)
plt.title('Distribution of age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Заменяем значение 'НЕТ ДЕТЕЙ МОЛОЖЕ 18' на 0
df['children'] = df['children'].replace('НЕТ ДЕТЕЙ МОЛОЖЕ 18', 0)

# Преобразуем колонку к числовому типу (если это необходимо)
df['children'] = pd.to_numeric(df['children'])
df['children']
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# Выбор нужных переменных для анализа корреляции
selected_variables = ['wage', 'work_hours', 'children', 'age']
selected_df = df[selected_variables]

# Построение матрицы корреляций
correlation_matrix = selected_df.corr()

# Визуализация матрицы корреляций с помощью тепловой карты
plt.figure(figsize=(8, 6))
heatmap = sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")
plt.title('Correlation Matrix')
plt.show()
```
